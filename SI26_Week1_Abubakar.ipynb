## Question 1

### What is OCR (Optical Character Recognition)?

**Answer:**

Optical Character Recognition (OCR) is a technology that converts printed or handwritten text from images into editable and searchable digital text. It helps computers recognize characters from scanned documents, books, or photographs. OCR saves time by reducing manual typing and makes documents easier to store, edit, and search.

---

## Question 2

### Why is Urdu OCR harder than English OCR?

**Answer:**

Urdu OCR is more difficult than English OCR because Urdu letters change their shape depending on their position in a word. Urdu is written from right to left, and many letters are connected, making text recognition more challenging. Different fonts, handwriting styles, and low-quality images also reduce OCR accuracy.

---

## Question 3

### What are two real-world situations where Urdu OCR would be useful?

**Answer:**

Urdu OCR is useful for converting old Urdu books and newspapers into digital text for easy searching and storage. It is also helpful in government offices and libraries to digitize Urdu documents and records. This technology saves time, reduces manual typing, and makes information easier to access.

```python
def preprocess_image(image_path, save_path):

    img = cv2.imread(image_path)

    if img is None:
        print(f"Could not load: {image_path}")
        return


    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


    resized = cv2.resize(gray, (512, 128))


    denoised = cv2.fastNlMeansDenoising(resized, h=10)


    _, binary = cv2.threshold(denoised, 127, 255, cv2.THRESH_BINARY)


    cv2.imwrite(save_path, binary)

    return binary


os.makedirs("data/processed", exist_ok=True)

print("Preprocessing function ready!")
```

```python
import glob
import os

# Find all images in data/raw/
all_images = glob.glob("data/raw/**/*.jpg", recursive=True)
all_images += glob.glob("data/raw/**/*.png", recursive=True)
all_images += glob.glob("data/raw/**/*.jpeg", recursive=True)

print(f"Found {len(all_images)} images to process")

processed_count = 0

for img_path in all_images:
    filename = os.path.basename(img_path)
    save_path = f"data/processed/{filename}"

    result = preprocess_image(img_path, save_path)

    if result is not None:
        processed_count += 1

print(f"Done! Processed {processed_count} images")
print("Check the data/processed folder.")
```

```python
from PIL import Image, ImageDraw, ImageFont
import arabic_reshaper
from bidi.algorithm import get_display
import os

# Create folder
os.makedirs("data/raw/synthetic", exist_ok=True)

# Urdu sentences (20)
urdu_texts = [
    "پاکستان زندہ باد",
    "آج کا موسم خوشگوار ہے",
    "تعلیم ہر انسان کا حق ہے",
    "کراچی پاکستان کا بڑا شہر ہے",
    "محنت کامیابی کی کنجی ہے",
    "میرا نام علی ہے",
    "اللہ سب سے بڑا ہے",
    "پاکستان ایک خوبصورت ملک ہے",
    "ہم سب کو محنت کرنی چاہیے",
    "علم روشنی ہے",
    "اسلام امن کا مذہب ہے",
    "زندگی ایک امتحان ہے",
    "صبح بخیر",
    "شکریہ",
    "آپ کیسے ہیں",
    "یہ ایک اردو جملہ ہے",
    "کامیابی مسلسل محنت سے ملتی ہے",
    "ہر دن کچھ نیا سیکھیں",
    "وقت بہت قیمتی ہے",
    "پاکستان ہمارا وطن ہے"
]

font_path = "NotoNaskhArabic-Regular.ttf"
font = ImageFont.truetype(font_path, 36)

for i, text in enumerate(urdu_texts, start=1):

    reshaped = arabic_reshaper.reshape(text)
    bidi_text = get_display(reshaped)

    img = Image.new("RGB", (700, 100), "white")
    draw = ImageDraw.Draw(img)

    draw.text((20,30), bidi_text, font=font, fill="black")

    img.save(f"data/raw/synthetic/urdu_{i}.png")

print("20 Synthetic Images Generated Successfully!")
```

```python
import pytesseract
from PIL import Image
import os
import pandas as pd

folder = "data/raw/newspaper"

rows = []

for file in os.listdir(folder):
    if file.lower().endswith((".png",".jpg",".jpeg")):
        text = pytesseract.image_to_string(
            Image.open(os.path.join(folder,file)),
            lang='eng'
        )

        rows.append({
            "image": file,
            "text": text.strip()
        })

df = pd.DataFrame(rows)
df.to_csv("screenshots_labels.csv", index=False)

print(df.head())
```

```python
import easyocr
import pandas as pd
import os

# Urdu + English OCR
reader = easyocr.Reader(['ur','en'])

folder = "/content/data/raw/newspaper"   # <-- apna folder path

data = []

for file in os.listdir(folder):
    if file.lower().endswith((".png", ".jpg", ".jpeg")):
        image_path = os.path.join(folder, file)

        result = reader.readtext(image_path, detail=0)

        text = " ".join(result)

        data.append({
            "image": file,
            "text": text
        })

df = pd.DataFrame(data)

df.to_csv("labels.csv", index=False, encoding="utf-8-sig")

print(df.head())
print("Total Images:", len(df))
```
