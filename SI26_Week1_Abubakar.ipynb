## Question 1

### What is OCR (Optical Character Recognition)?

**Answer:**

Optical Character Recognition (OCR) is a technology that converts printed or handwritten text from images into editable and searchable digital text. It helps computers recognize characters from scanned documents, books, or photographs. OCR saves time by reducing manual typing and makes documents easier to store, edit, and search.

## Question 2

### Why is Urdu OCR harder than English OCR?

**Answer:**

Urdu OCR is more difficult than English OCR because Urdu letters change their shape depending on their position in a word. Urdu is written from right to left, and many letters are connected, making text recognition more challenging. Different fonts, handwriting styles, and low-quality images also reduce OCR accuracy.

## Question 3

### What are two real-world situations where Urdu OCR would be useful?

**Answer:**

Urdu OCR is useful for converting old Urdu books and newspapers into digital text for easy searching and storage. It is also helpful in government offices and libraries to digitize Urdu documents and records. This technology saves time, reduces manual typing, and makes information easier to access.
